```python
import random
import string

# Define a function to generate a random string of letters
def generate_name(length):
    letters = string.ascii_lowercase
    return ''.join(random.choice(letters) for i in range(length))

# Define a list of domain names
domains = ['gmail.com', 'yahoo.com', 'hotmail.com', 'outlook.com']

# Loop 100 times to generate 100 email addresses
for i in range(100):
    
    # Generate a first name with length between 5 and 10 letters
    first_name = generate_name(random.randint(5, 10))
    
    # Generate a last name with length between 5 and 10 letters
    last_name = generate_name(random.randint(5, 10))
    
    # Combine the first and last name with a period
    name = first_name + '.' + last_name
    
    # Choose a random domain name from the list of domains
    domain = random.choice(domains)
    
    # Generate the email address with the name and domain
    email = name + '@' + domain
    
    # Append the email address to the list of emails
    emails.append(email)

# Print the list of email addresses
# print(emails)
```
